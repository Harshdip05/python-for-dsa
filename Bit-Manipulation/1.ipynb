### Binary AND

Only when both bits are 1 → result is 1

Otherwise → 0

```python
5 & 6
```

```python
print(5 & 6)
```

### Binary OR

If ANY bit is 1 → result is 1

```python
print(5 | 6)
```

### Binary XOR

Same bits → 0

Different bits → 1

```python
5 ^ 6
```

### Check even or odd

```python
def even_odd(n):
    if (n & 1) == 1:
        return "odd"
    else:
        return "even"
    
print(even_odd(10))  
print(even_odd(7))


def even_odd(n):
    if n & 1 :
        return "odd"
    else:
        return "even"
    
print(even_odd(10))  
print(even_odd(7))
```

```python
def even_odd(n):
    if (n & 1) == 0:
        return "odd"
    else:
        return "even"
    
print(even_odd(10))  
print(even_odd(7))

def even_odd(n):
    if n & 1 :
        return "odd"
    else:
        return "even"
    
print(even_odd(10))  
print(even_odd(7))
```

```python
def even_odd(n):
    if n & 1 == 0:   ##   n & 0  → always 0  👉 So condition is always false
        return "odd"
    else:
        return "even"
    
print(even_odd(10))  
print(even_odd(7))
```

```python
def get_ith_bit(n, i):
    return (n >> i) & 1

print(get_ith_bit(10, 1))  # 1
print(get_ith_bit(10, 2))  # 0
```

```python
def get_ith_bit(n, i):
    if n & (1 << i):
        return 1
    else:
        return 0
```
