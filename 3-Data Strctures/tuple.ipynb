| Feature | Tuple  | List   |
| ------- | ------ | ------ |
| Mutable | ❌ No   | ✅ Yes  |
| Speed   | Faster | Slower |
| Methods | Few    | Many   |
| Syntax  | `( )`  | `[ ]`  |


| Feature | Packing        | Unpacking      |
| ------- | -------------- | -------------- |
| Meaning | Combine values | Extract values |
| Example | `t = 1,2,3`    | `a,b,c = t`    |

```python
# Packing = putting multiple values into a single variable (tuple by default)
t = 1, 2, 3
print(t)
# Python automatically packs values into a tuple
```

```python
t = (1, 2, 3)    ## With parentheses (optional)
print(t)
```

```python
data = "Aman", 20, "India"
print(data)
```

```python
# Unpacking = extracting values from a tuple (or list) into variables
t = (1, 2, 3)

a, b, c = t

print(a)  # 1
print(b)  # 2
print(c)  # 3
```

```python
# Number of variables must match number of values
t = (1, 2, 3)

a, b = t   # ❌ Error
```

```python
t = (1, 2, 3, 4, 5)

a, *b = t

print(a)  # 1
print(b)  # [2, 3, 4, 5]
```

```python
a, *b, c = (1, 2, 3, 4, 5)

print(a)  # 1
print(b)  # [2, 3, 4]
print(c)  # 5
```

```python
# Swapping variables
a = 5
b = 10

a, b = b, a

print(a, b)
# no temporary variable needed
```

```python
t = (1, 2, 2, 3)
print(t.count(2))

t = (10, 20, 30)
print(t.index(20))
```
