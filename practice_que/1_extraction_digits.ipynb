```python
n = 5873
digits = []

while n > 0:
    digit = n % 10
    digits.append(digit)
    n = n // 10

print(digits)
```

-- Common examples:

O(1) → constant work

O(log n) → repeatedly divide by 2, 10, etc.

O(n) → go through n elements once

O(n log n) → n elements × log n work

O(n²) → nested loops, each running n times



-- So remember:

One loop that divides n → O(log n)

One loop that runs n times → O(n)

Nested n loop + log n loop → O(n log n)




Complexity	Result:

Time	O(log n)

Space	O(1)

If a loop repeatedly does n = n // 10, you can usually recognize the time complexity as O(log n) because you're reducing the number by a factor of 10 each iteration.
