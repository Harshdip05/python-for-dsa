##### “Why is it called bubble sort?”


👉
“Because in each pass, the largest element moves step by step through adjacent swaps and rises to its correct position, 

just like a bubble rising to the surface.”

```python
def bubble_sort(arr):
    n = len(arr)

    for j  in range(0,n-1):
        if arr[j]>arr[j+1]:
            arr[j],arr[j+1] = arr[j+1],arr[j]
    return arr        

unsorted_list = [12,25,11,34,90,22]
sorted_list = bubble_sort(unsorted_list)
print("Sorted elements: ",sorted_list)
```

| Sorting Type | Condition |
