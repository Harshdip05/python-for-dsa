```python
def third_largest(arr):
    if len(arr) < 3:
        return -1

    first = second = third = float('-inf')

    for num in arr:
        if num > first:
            third = second
            second = first
            first = num

        elif num > second and num != first:
            third = second
            second = num

        elif num > third and num != second and num != first:
            third = num

    return third if third != float('-inf') else -1
```

```python
# Instead of comparing every element with both min and max, I compare elements in pairs. This reduces comparisons from 2n to about 3n/2.

# pair comparsion

def min_max_pair(arr):
    if not arr:
        return None, None

    n = len(arr)

    # Initialize
    if n % 2 == 0:
        if arr[0] < arr[1]:
            min_val, max_val = arr[0], arr[1]
        else:
            min_val, max_val = arr[1], arr[0]
        i = 2
    else:
        min_val = max_val = arr[0]
        i = 1

    # Process pairs
    while i < n - 1:
        if arr[i] < arr[i+1]:
            small, large = arr[i], arr[i+1]
        else:
            small, large = arr[i+1], arr[i]

        if small < min_val:
            min_val = small
        if large > max_val:
            max_val = large

        i += 2

    return min_val, max_val
```

## Binary Search

used to find an element in a sorted array.
Instead of checking elements one by one like linear search, it divides the array into halves repeatedly.

Because of this, it is much faster for large datasets.

```python
# Duplicate Elements
# arr = [10, 20, 20, 20, 30]

# 👉 May return any occurrence
```

```python
def search(nums, target):
    start = 0
    end = len(nums) - 1

    while start <= end:
        mid = (start + end) // 2       ## mid should be updated every time the search range changes.

        if nums[mid] == target:        ## But mid is an index, not the value.
            return mid
        elif nums[mid] < target:
            start = mid + 1
        else:
            end = mid - 1

    return -1

nums = [-1, 0, 3, 5, 9, 12]
target = 9

print(search(nums, target))

# mid changes every iteration So it must be inside the loop.

# You don’t know how many guesses needed So you use while, not for.
# Number of iterations is not fixed
# Search space changes dynamically
# Loop stops based on condition
```

```python
# recursive code

def binary_recursive(arr,start,end,target):
    
    if start>end:
        return -1
    
    mid = (start+end)//2

    if arr[mid] == target:
        return mid
    elif mid>target:
        return binary_recursive(arr,start,end,target)
    else:
        return binary_recursive(arr,start,end,target)
    

# | Type      | Time     | Space    |
# | --------- | -------- | -------- |
# | Iterative | O(log n) | O(1)     |
# | Recursive | O(log n) | O(log n) |    ## Each function call is stored in call stack
```

```python
def binary_search(arr, low, high, target):
    if low > high:
        return -1

    mid = (low + high) // 2

    if arr[mid] == target:
        return mid
    elif arr[mid] < target:
        return binary_search(arr, mid+1, high, target)
    else:
        return binary_search(arr, low, mid-1, target)
```

```python
def first_occurrence(arr, target):
    left, right = 0, len(arr) - 1
    result = -1

    while left <= right:
        mid = (left + right) // 2

        if arr[mid] == target:
            result = mid
            right = mid - 1   # go LEFT
        elif arr[mid] < target:
            left = mid + 1
        else:
            right = mid - 1

    return result
```

## Reverse an Array
Reversing an array means changing the order of elements from end to start

| Method           | Time | Space |
| ---------------- | ---- | ----- |
| New array copy   | O(n) | O(n)  |
| Two pointer swap | O(n) | O(1)  |


| Feature    | Brute Force | Two Pointer 🔥 |
| ---------- | ----------- | -------------- |
| Time       | O(n)        | O(n)           |
| Space      | O(n) ❌      | O(1) ✅         |
| Efficiency | Less        | Best           |
| Interview  | Basic       | ⭐⭐⭐⭐⭐          |


```python
# Brute force uses extra space by creating a new array.
# A better approach is two pointers, swapping elements in-place with O(1) space

def reverse_array(arr):
    n = len(arr)
    result = []
    
    for i in range(n - 1, -1, -1):
        result.append(arr[i])
    
    return result


arr = [1, 2, 3, 4, 5]
print(reverse_array(arr))
```

```python
def reverse(arr):
    left = 0
    right = len(arr)-1
    while(left<right):

        arr[left],arr[right] = arr[right],arr[left]
        left +=1
        right -=1

    return arr

arr = [1,2,3,4,5]
print(reverse(arr))

# left = 0 → means index (position)
# arr[0] → means value at that position
# 👉 Two-pointer always works with indices, not values
```

```python
def reverse_array(arr):
    left = 0
    right = len(arr) - 1

    while left < right:
        temp = arr[left]
        arr[left] = arr[right]
        arr[right] = temp

        left += 1
        right -= 1

    return arr


arr = [1, 2, 3, 4, 5]
print(reverse_array(arr))
```

```python
def reverse_array(arr):
    new_arr = []

    for x in arr[::-1]:
        new_arr.append(x)

    return new_arr


arr = [1,2,3,4,5]
print(reverse_array(arr))
```

Use reverse() when: You want to modify the same list

Use reversed() when: You want a reversed version without changing original data

| Feature               | `reverse()` | `reversed()`                 |

## Pairs in an array

A pair means two elements taken together from an array.

Example array:
[1, 2, 3]

Possible pairs:
(1,2)
(1,3)
(2,3)

Each pair contains two different elements.

### If array size = n          
### Total pairs: n(n-1)/2

```python
# Brute Force (2 Loops)
def print_pairs(arr):
    for i in range(len(arr)):
        # print("first loop index is ",i)
        for j in range(i+1, len(arr)):
            print(arr[i], arr[j])


arr = [1,2,3,4]
print_pairs(arr)
```

```python
def print_pairs(arr):
    n = len(arr)

    for i in range(n):
        for j in range(i + 1, n):
            print(arr[i], arr[j])


arr = [1, 2, 3]
print_pairs(arr)
```

```python
# Store Pairs in List
def get_pairs(arr):
    pairs = []
    total_pairs = 0

    for i in range(len(arr)):
        for j in range(i+1, len(arr)):
            pairs.append((arr[i], arr[j]))
            total_pairs +=1

    # return pairs
    return total_pairs
arr = [1,2,3,4]
get_pairs(arr)    
```

## Max Subarray Sum

| Method                | Loops   | Complexity |
| --------------------- | ------- | ---------- |
| Basic brute force     | 3 loops | O(n³)      |
| Optimized brute force | 2 loops | O(n²)      |
| Kadane's Algorithm    | 1 loop  | O(n)       |


```python
arr = [-2,1,-3,4,-1,2,1,-5,4]

current_sum = 0
max_sum = float('-inf')

for num in arr:
    current_sum += num
    
    if current_sum > max_sum:
        max_sum = current_sum
    
    if current_sum < 0:
        current_sum = 0

print("Maximum subarray sum:", max_sum)
```

```python
def stock(arr):

    max_profit = 0  ## if no profit return 0

    for i in range(len(arr)):
        for j in range(i+1,len(arr)):

            if arr[j]>arr[i]:
                profit = arr[j] - arr[i]

                if profit>max_profit:
                    max_profit = profit

    return max_profit

arr = [3,1,2,3,4,5,6]
print(stock(arr))                    
```

```python
# two pointer
# I use two pointers where one tracks the minimum price so far, and the other scans for maximum profit in one pass
def max_profit(prices):
    left = 0   # buy day
    right = 1  # sell day
    max_profit = 0

    while right < len(prices):
        if prices[right] > prices[left]:
            profit = prices[right] - prices[left]
            max_profit = max(max_profit, profit)
        else:
            left = right   # found smaller price → better buy

        right += 1

    return max_profit

```
