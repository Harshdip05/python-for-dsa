```python
rows = 3
cols = 3

matrix = []

for i in range(rows):
    row = []
    for j in range(cols):
        row.append(0)
    matrix.append(row)


print(matrix)
```

```python
matrix = [[0 for j in range(3)] for i in range(3)]
print(matrix)
```

```python
for i in range(len(matrix)):
    for j in range(len(matrix[0])):
        print(matrix[i][j])
```

```python
for row in matrix:
    for val in row:
        print(val)
```
