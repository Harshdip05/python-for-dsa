```python
# It returns the memory size (in bytes) of an object.

import sys

x = 10
print(sys.getsizeof(x))
```

```python
import sys

x = [1,2,3]
print(sys.getsizeof(x))
```

```python
import sys

print(sys.getsizeof(10))           # integer
print(sys.getsizeof(3.14))        # float
print(sys.getsizeof("hello"))     # string
print(sys.getsizeof([1,2,3]))     # list
```

| Method       | Stored value |
| ------------ | ------------ |
| `input()`    | `"hello"`    |
| `readline()` | `"hello\n"`  |


```python
s = input()
print(s)
print(len(s))
```

```python
import sys

s = sys.stdin.readline()
print(s)
print(len(s))

# Why length = 6?
# Because Python sees: "h e l l o \n"
# That’s 6 characters.
```

```python
# Remove newline using .strip()

import sys

s = sys.stdin.readline().strip()
print(s)
print(len(s))
```

```python
import sys
input = sys.stdin.readline

s = input().strip()
if s == "hello":
    print("YES")
```
