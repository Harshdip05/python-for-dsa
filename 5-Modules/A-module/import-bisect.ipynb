```python
# bisect is a Python module that uses binary search to find insertion positions in a sorted list
import bisect
# Find where to insert an element
# Insert element in correct sorted position
```

| Function       | Meaning                      |
| -------------- | ---------------------------- |
| `bisect_left`  | Insert **before duplicates** |
| `bisect_right` | Insert **after duplicates**  |

| Case                | bisect_left | bisect_right |
| ------------------- | ----------- | ------------ |
| Element not present | Same        | Same         |
| Element present     | First pos   | Last pos +1  |



### bisect_left(arr, x)

### bisect_right(arr, x)

```python
# left → go to first possible place
# right → go to last possible place

import bisect

arr = [10, 20, 30, 40]
pos = bisect.bisect_left(arr, 25)

print(pos)



import bisect

arr = [10, 20, 30, 40]
pos = bisect.bisect_right(arr, 25)

print(pos)
```

| Concept     | Condition | Meaning                    |
| ----------- | --------- | -------------------------- |
| Lower Bound | ≥ target  | First valid position       |
| Upper Bound | > target  | Next position after target |


```python
# Lower bound finds the first element ≥ target, 
# while upper bound finds the first element strictly greater than target.

```

```python
# Given a sorted array, count how many times a number appears.

import bisect

def count_frequency(arr, target):
    left = bisect.bisect_left(arr, target)
    right = bisect.bisect_right(arr, target)
    
    return right - left

arr = [2, 4, 4, 4, 6]
target = 4    
print(count_frequency(arr,target))


from collections import Counter
count = Counter(arr)
print(count)
print(count[4])
```
