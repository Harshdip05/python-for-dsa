```python
def linear_search(arr,target):
    for i in range(len(arr)):
        if arr[i] == target:
            return i
    return -1    

linear_search([1,2,3,4,],5)        
```

```python
def binary_search(arr,target):
    first = 0
    last = len(arr)-1

    while  first<=last:
        mid = (first+last)//2
        if mid == target:
            return mid
        elif mid<target:
            first = mid + 1 
        else:
            last = mid - 1
    return -1
```
