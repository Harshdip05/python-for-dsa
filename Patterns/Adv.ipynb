#### hollow rectangle pattern

```python
n = 4
m = 5

for i in range(n):
    for j in range(m):
        print("*",end=" ")
    print()        
```

```python
n = 4
m = 5

for i in range(n):
    for j in range(m):
        if i == 0 or i == n-1 or j == 0 or j == m-1:
            print("*",end=" ")
        else:
            print(" ",end =" ")      
    print()           
```

```python
# Inverted Half Pyramid

n = 4

for i in range(4):
    for j in range(n-i):
        print("*",end = " ")
    print()    
```

```python
# Rotated Half Pyramid (Right-Aligned)

n = 4
for i in range(n):
    # spaces
    for j in range(n - i - 1):
        print(" ", end=" ")
    
    # stars
    for j in range(i + 1):
        print("*", end=" ")
    
    print()
```

```python
n = 4

for i in range(n):
    # spaces (increase)
    for j in range(i):
        print(" ", end=" ")

    # stars (decrease)
    for j in range(n - i):
        print("*", end=" ")

    print()
```

```python
# butterfly pattern

n = 4

for i in range(0,n):
    
    for j in range(i+1):
        print("*",end="")
    for j in range(0,2*(n-i)):
        print(" ",end="")    
    for j in range(i+1):
        print("*",end="")   

    print()        

for i in range(n,-1,-1):
    
    for j in range(i+1):
        print("*",end="")
    for j in range(0,2*(n-i)):
        print(" ",end="")    
    for j in range(i+1):
        print("*",end="")   

    print()   
```

```python
n = 4

# upper half
for i in range(n):
    # left stars
    for j in range(i+1):
        print("*", end="")
    
    # spaces
    for j in range(2*(n-i-1)):
        print(" ", end="")
    
    # right stars
    for j in range(i+1):
        print("*", end="")
    
    print()

# lower half
for i in range(n):
    # left stars
    for j in range(n-i):
        print("*", end="")
    
    # spaces
    for j in range(2*i):
        print(" ", end="")
    
    # right stars
    for j in range(n-i):
        print("*", end="")
    
    print()
```
