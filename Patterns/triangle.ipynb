```python
### star pattern

n = 4

for i in range(n):
    for j in range(i+1):
        print("*",end=" ")
    print()    
```

```python
n = 4

for i in range(n):
    for j in range(n):
        print("*"*(j+1))
    break




n = 4

for i in range(n):
    for j in range(n):
        print("*"*(j+1),end ="\n")
    break
```

```python
###  inverted star pattern

n = 4
for i in range(n):
    for j in range(n-i-1,-1,-1):
        print("*",end=" ")
    print()    
```

```python
n = 4

for i in range(n):
    for j in range(n-i):
        print("*",end =" ")
    print()    
```

```python
### half pyramid pattern


n = 4

for i in range(n):
    for j in range(i+1):
        print(j+1,end=" ")
    print()   
```

```python
for i in range(n):
    for j in range(i+1):
        print(i+1, end=" ")
    print()
```
