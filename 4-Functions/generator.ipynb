```python
# A generator is an object that:

# Produces values one at a time (lazy evaluation)
# Remembers its state between calls
# Uses yield instead of return
```

### Generator

A generator expression in Python is a simple and memory-efficient way to create iterators.


| Feature | List comprehension    | Generator expression    |
| ------- | --------------------- | ----------------------- |
| Syntax  | `[x for x in ...]`    | `(x for x in ...)`      |
| Memory  | Stores all values     | Generates one at a time |
| Speed   | Faster for small data | Better for large data   |



| Feature     | Generator Function       | Generator Expression   |
| ----------- | ------------------------ | ---------------------- |
| Syntax      | `def + yield`            | `(x for x in ...)`     |
| Length      | Multi-line               | One-line               |
| Complexity  | Can handle complex logic | Best for simple logic  |
| Readability | Better for big logic     | Better for short tasks |


```python
squares = (x**2 for x in [1,2,3,4])
print(squares)                               ## generator object
print(id(squares))                           ## the unique memory address (identity) of that generator object
```

```python
nums =  [1,2,3,4]
squares = (x**2 for x in nums)
print(squares)

# It does NOT compute immediately.
```

### How to use a generator

```python
## method 1 : loop

squares = (x**2 for x in nums)
for x in squares:
    print(x)
```

```python
## method 2 : next()

squares = (x**2 for x in [1,2,3,4])

print(next(squares))  # 1
print(next(squares))  # 4
print(next(squares))  # 9
print(next(squares))
# print(next(squares))       ## stop iteration
```

```python
## method 3 : list

squares = (x**2 for x in nums)
print(list(squares))
```

```python

squares = (x**2 for x in nums)
print(tuple(squares))
```

### Important behaviour

```python
g = (x for x in [1,2,3])

for i in g:
    print(i)

for i in g:
    print(i)  # Nothing prints (already exhausted)
```

### Generator expression

(x for x in iterable)

👉 It generates values one by one, not all at once.

```python
gen = (x for x in range(10) if x % 2 == 0)

for i in gen:
    print(i)
```

```python
gen = ("Even" if x % 2 == 0 else "Odd" for x in range(5))


for i in gen:
    print(i)
```

```python
gen = (x*y for x in [1,2,3] for y in [10,20])    ## multiple loops


for i in gen:
    print(i)
```

```python
gen = (x for x in range(20) if x % 2 == 0 if x > 10)  ##3 nested condition


for i in gen:
    print(i)
```

```python
gen = (len(word) for word in ["hi", "hello", "python"])


for i in gen:
    print(i)
```

```python
total = sum(x for x in range(5))
print(total)

# sum() with generator expression computes values lazily without creating a list, making it memory efficient.
```

```python
total = sum(x*x for x in range(5))
print(total)
```

```python
total = sum(x for x in range(10) if x % 2 == 0)
print(total)
```

### Iterable VS Iterator

| Feature       | Iterable     | Iterator             |
| ------------- | ------------ | -------------------- |
| What it is    | A collection | A value producer     |
| Function used | `iter()`     | `next()`             |
| Stores data   | Yes          | No (just pointer)    |
| Examples      | list, tuple  | generator, iter(obj) |


```python
# An iterable is something you can loop over

list = [1, 2, 3]
for x in list:
    print(x)             ## list is an iterable
```

```python
# An iterator is an object that:
            # Keeps track of current position
            # Gives next value using next()


it = iter([1, 2, 3])

print(next(it))  # 1
print(next(it))  # 2
print(next(it))  # 3

# [1,2,3] → iterable
# it → iterator
```
