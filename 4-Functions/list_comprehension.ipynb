### List Comprehension

```python
# [expression for item in iterable if condition]

# expression → what you want to put in the list
# item → each element from the iterable
# iterable → list, tuple, string, etc.
# condition (optional) → filter items
```

```python
squares = []
for i in range(5):
    squares.append(i*i)   ## without list comprehension

print(squares)    
```

```python
squares = [i*i for i in range(5)]
print(squares)
```

```python
evens = [i for i in range(10) if i % 2 == 0]
print(evens)
```

```python
result = ["Even" if i % 2 == 0 else "Odd" for i in range(5)]
print(result)
```

```python
letters = [char.upper() for char in "hello"]
print(letters)
```

```python
matrix = [[j for j in range(3)] for i in range(3)]
print(matrix)
```
