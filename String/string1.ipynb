### String Concatenation 

String concatenation means joining two or more strings together to form a single string


Concatenating strings usually takes:

O(n)

Where n = length of the resulting string.

Repeated concatenation in loops can become O(n²) in some languages.

```python
s1 = "Hello"
s2 = "World"

result = s1 + s2

print(result)
```

```python
s1 = "Hello"
s2 = "World"

result = s1 + " " + s2

print(result)
```

```python
## Concatenation Using Loop
# You can concatenate multiple strings.

words = ["I", "love", "coding"]

result = ""

for word in words:
    result += word + " "

print(result)
```

### Palindrome Check


```python
s = input()
# x = input()

def palindrome(s):
    if s == s[::-1]:
        return True
    else:
        return False  

s1 = palindrome(s)
print(s1)          
```

Two strings are anagrams if they contain the same characters in different order.

Example:

listen
silent

Both have same letters.

```python
# Starting from (0,0), find the shortest distance from the origin after following the path.

# N = y + 1
# S = y - 1
# W = x - 1
# E = x + 1

import math

path = input()

x = 0
y = 0

for direction in path:
    if direction == 'N':
        y += 1
    elif direction == 'S':
        y -= 1
    elif direction == 'E':
        x += 1
    elif direction == 'W':
        x -= 1

distance = math.sqrt(x*x + y*y)

print(distance)
```

### String interning 

Storing only one copy of identical strings in memory to save memory and improve performance.

If two variables have the same string, they may point to the same memory location.

```python
# In Python, small or commonly used strings are automatically interned.

a = "hello"
b = "hello"

print(a is b)
```

is checks memory identity, not value.

This means both variables reference the same object.

```python
a = "hello world"
b = "hello world"

print(a is b)   ## IDs are different → two separate objects.
print(id(a))
print(id(b))
```

```python
# Forcing String Interning
# Python provides a function to explicitly intern strings

import sys

a = sys.intern("hello world")
b = sys.intern("hello world")

print(a is b)  ## Now both variables refer to the same interned string.

print(id(a))
print(id(b))
```

| Operator | Meaning                   |
| -------- | ------------------------- |
| `==`     | compares values           |
| `is`     | compares memory addresses |


```python
a = "hello"
b = "hello"

print(a == b)   # True
print(a is b)   # True
```
