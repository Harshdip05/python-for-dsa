### flush

```python
import time

print("Loading...", end="",flush=False)
time.sleep(3)
print(" Done!")

# "Loading..." may NOT appear immediately
# It might show after 3 seconds together with "Done!"
```

```python
# flush=True forces Python to immediately write output instead of waiting.
import time

print("Loading...", end="", flush=True)
time.sleep(3)
print(" Done!")

# "Loading..." appears instantly
# After 3 seconds → " Done!" appears
```

### Static vs Dynamic

| Feature          | Static Typing    | Dynamic Typing   |
| ---------------- | ---------------- | ---------------- |
| Type checking    | Before execution | During execution |
| Flexibility      | Low              | High             |
| Error detection  | Early            | Late             |
| Speed            | Faster           | Slightly slower  |
| Example language | Java             | Python           |


```python
## Static Typing

        # Type is checked before running the program (at compile time)
        # 📌 Example (like C, Java)
        # int x = 10
        # x = "hello"   # ❌ Error
        # 👉 You cannot change type once defined


## Dynamic Typing

        # Type is checked while the program runs (runtime)
        # 📌 Example (Python)
        # x = 10
        # x = "hello"   # ✅ Allowed
        # 👉 Same variable can hold different types

def add(x: int, y: int) -> int:   ## python gives type hints, This doesn’t enforce types at runtime, but helps tools & readability
    return x + y        
```

| Feature      | Static Binding       | Dynamic Binding   |
| ------------ | -------------------- | ----------------- |
| When decided | Compile time         | Runtime           |
| Speed        | Faster               | Slightly slower   |
| Flexibility  | Less                 | More              |
| Example      | Function overloading | Method overriding |


```python
# Static Binding (Early Binding)
# Binding happens at compile time (before program runs)
```

```python
# Dynamic Binding (Late Binding)
# Binding happens at runtime (while program runs)
```

```python
# Python mostly uses dynamic binding
# Method calls are resolved at runtime
# That’s why Python is very flexible
```

### Keywords and identifiers

```python
# import keyword → gives access to Python’s reserved words
# keyword.kwlist → list of keywords
# keyword.iskeyword() → check if word is reserved
```

```python
import keyword

print(keyword.kwlist)

print(keyword.iskeyword("for"))   
print(keyword.iskeyword("hello"))
```

```python
# isidentifier() checks:  “Can this string be a valid Python variable name?"

print("x".isidentifier())        # True
print("my_var".isidentifier())   # True
print("_name".isidentifier())    # True
print("var123".isidentifier())   # True
print("1var".isidentifier())     # False
print("my var".isidentifier())   # False
print("var!".isidentifier())     # False
print("".isidentifier())         # False
```

### Literal

A literal is: A fixed value written directly in the code

```python
a = 0b1010 #Binary Literals
b = 100 #Decimal Literal 
c = 0o310 #Octal Literal
d = 0x12c #Hexadecimal Literal

#Float Literal
float_1 = 10.5 
float_2 = 1.5e2 # 1.5 * 10^2
float_3 = 1.5e-3 # 1.5 * 10^-3

#Complex Literal 
x = 3.14j

print(a, b, c, d)
print(float_1, float_2,float_3)
print(x, x.imag, x.real)
```
