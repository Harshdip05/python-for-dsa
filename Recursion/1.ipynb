Recursion by definition requires a function to call itself. Without a function, there’s nothing that can “call itself,” so recursion cannot exist.

Recursion → cleaner, easier to understand for problems like trees, graphs

Loops → usually faster and use less memory

#### When to Use Recursion
Tree or graph traversal

Divide and conquer (merge sort, quicksort)

Problems that can be broken into similar smaller problems

```python
# Factorial without recursion

n = 5
result = 1

for i in range(1, n + 1):
    result = result * i

print(result)   
```
